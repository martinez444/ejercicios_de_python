```python
# 1. Pedimos los datos al usuario y los convertimos a números (float)
numer1 = input(input("Inserta el numero 1: "))  # Solicita el primer número y lo convierte a decimal.
numer2 = input(input("Inserta el numero 2: "))  # Solicita el segundo número y lo convierte a decimal.


# 2. Definición de funciones

def sum(numer1, numer2):  # Definimos la función 'sum' que recibe numer1 y numer2.
    result = numer1 + numer2  # Sumamos los dos números y guardamos el resultado en 'result'.
    print(numer1, "+", numer2, "=", result)  # Imprimimos la operación y su resultado.

def sub(numer1, numer2):  # Definimos la función 'sub' para restar.
    result = numer1 - numer2  # Restamos numer2 a numer1 y lo guardamos en 'result'.
    print(numer1, "-", numer2, "=", result)  # Mostramos el resultado de la resta.

def mult(numer1, numer2):  # Definimos la función 'mult' para multiplicar.
    result = numer1 * numer2  # Multiplicamos ambos números y guardamos el producto en 'result'.
    print(numer1, "*", numer2, "=", result)  # Imprimimos la multiplicación.

def div(numer1, numer2):  # Definimos la función 'div' para dividir.
    if numer2 != 0:  # Validamos que numer2 no sea cero para evitar un error de división por cero.
        result = numer1 / numer2  # Dividimos numer1 entre numer2 si la condición es cierta.
        print(numer1, "/", numer2, "=", result)  # Imprimimos el resultado.
    else:  # Si numer2 es cero, se ejecuta este bloque.
        print("El Numero 2 es igual a 0 y no se puede dividir por 0.")  # Muestra mensaje de advertencia.


# 3. Llamada a las funciones

sum(numer1, numer2)  # Ejecuta la función de suma pasando las variables capturadas arriba.
sub(numer1, numer2)  # Ejecuta la función de resta.
mult(numer1, numer2)  # Ejecuta la función de multiplicación.
div(numer1, numer2)  # Ejecuta la función de división.


```
