```python
inventario = 100
print (f"El inventario contiene {inventario} hamburguesas")
while inventario > 0:
  if inventario <=0:
      print("Advertencia: El inventario esta casi vacio")
  num_hamburguesas = int(input("Cuantas hamburguesas quiere el cliente"))
  if num_hamburguesas > inventario:
       print("No hay sufientes hamburguesas en stock. solo hay", inventario, "hamburguesas disponibles")
  else:
    inventario-= num_hamburguesas
    print(f"El cliente ha pedido {num_hamburguesas} hamburguesas. El inventario ahora tiene tiene {inventario} hamburguesas" )
```
