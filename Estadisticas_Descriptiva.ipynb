📢 Historia:
Un grupo de agricultores está participando en un concurso para ver quién tiene las zanahorias más grandes. Han medido el peso (en gramos) de 12 zanahorias y necesitan analizar los datos para ver qué tan uniformes son sus cultivos.
📌 Tu misión
📊 Escribe un programa en Python que haga lo siguiente:
Calcule la media (promedio) del peso de las zanahorias.
Encuentre la mediana del peso.
Determine la moda (peso más común).
Calcule la desviación estándar para ver qué tanto varían los pesos.
Muestre los resultados de forma clara y divertida.
📌 Estos son los pesos (en gramos) de las zanahorias concursantes:
zanahorias = [300, 280, 290, 310, 275, 290, 295, 315, 290, 280, 310, 305]

🚀 Bonus (Opcional)
📊 Extra 1: Agrega una visualización con Matplotlib 📉 para mostrar la distribución de los pesos.
🤖 Extra 2: Modifica el código para que el usuario pueda ingresar los pesos de nuevas zanahorias.

```python

# ==========================================
# CONCURSO DE LAS ZANAHORIAS MÁS GRANDES
# ==========================================

# Importamos las librerías necesarias
import numpy as np
import statistics as st
import matplotlib.pyplot as plt

# Lista con los pesos de las 12 zanahorias (en gramos)
zanahorias = [300, 280, 290, 310, 275, 290,
              295, 315, 290, 280, 310, 305]

# 1. CALCULAR LA MEDIA (PROMEDIO)
media = np.mean(zanahorias)

# 2. CALCULAR LA MEDIANA (VALOR CENTRAL)
mediana = np.median(zanahorias)

# 3. CALCULAR LA MODA (VALOR MÁS REPETIDO)
moda = st.mode(zanahorias)

# 4. CALCULAR LA DESVIACIÓN ESTÁNDAR
desviacion = np.std(zanahorias)

# 5. MOSTRAR LOS RESULTADOS
print("¡RESULTADOS DEL CONCURSO DE ZANAHORIAS!")
print("=" * 45)

print(f"Media: {media:.2f} gramos")
print(f"Mediana: {mediana:.2f} gramos")
print(f"Moda: {moda} gramos")
print(f"Desviación estándar: {desviacion:.2f} gramos")

print("=" * 45)
print("¡Análisis de la cosecha terminado!")

# 6. VISUALIZACIÓN CON MATPLOTLIB
plt.figure(figsize=(9, 5))

# Crear un histograma con los pesos
plt.hist(zanahorias, bins=6, edgecolor="black")

# Añadir título y etiquetas
plt.title("Distribución de los pesos de las zanahorias")
plt.xlabel("Peso en gramos")
plt.ylabel("Número de zanahorias")

# Mostrar la cuadrícula
plt.grid(axis="y", linestyle="--", alpha=0.5)

# Mostrar el gráfico
plt.show()
```

```python

# ==========================================
# 🥕 CONCURSO DE LAS ZANAHORIAS
# ==========================================

# Importamos las librerías necesarias
import numpy as np
import statistics as st
import matplotlib.pyplot as plt

# Creamos una lista vacía para guardar los pesos
zanahorias = []

# Mostramos las instrucciones
print("🥕 ¡Bienvenido al concurso de zanahorias!")
print("Introduce el peso de cada zanahoria en gramos.")
print("Escribe 'salir' cuando quieras terminar.")
print("-" * 45)

# Repetimos la entrada hasta que el usuario escriba salir
while True:

    # Pedimos el peso de una nueva zanahoria
    entrada = input("Introduce el peso de una zanahoria: ")

    # Comprobamos si el usuario quiere terminar
    if entrada.lower() == "salir":
        break

    # Intentamos convertir el dato en un número
    try:
        peso = float(entrada)

        # Comprobamos que el peso sea positivo
        if peso <= 0:
            print("⚠️ El peso debe ser mayor que cero.")
            continue

        # Guardamos el peso en la lista
        zanahorias.append(peso)

        # Confirmamos que se ha añadido
        print(f"✅ Zanahoria de {peso:g} gramos añadida.")

    except ValueError:
        # Avisamos si el dato no es un número válido
        print("⚠️ Introduce un número válido o escribe 'salir'.")

# Comprobamos si se han introducido zanahorias
if len(zanahorias) > 0:

    # Calculamos las estadísticas
    media = np.mean(zanahorias)
    mediana = np.median(zanahorias)
    moda = st.mode(zanahorias)
    desviacion = np.std(zanahorias)

    # Mostramos los resultados
    print("\n🥕 RESULTADOS DEL CONCURSO 🥕")
    print("=" * 40)
    print(f"Cantidad de zanahorias: {len(zanahorias)}")
    print(f"📊 Media: {media:.2f} gramos")
    print(f"🎯 Mediana: {mediana:.2f} gramos")
    print(f"🏆 Moda: {moda:g} gramos")
    print(f"📉 Desviación estándar: {desviacion:.2f} gramos")
    print(f"📋 Pesos registrados: {zanahorias}")
    print("=" * 40)

    # Creamos el gráfico
    plt.figure(figsize=(9, 5))
    plt.hist(zanahorias, bins=6, edgecolor="black")
    plt.title("Distribución de los pesos de las zanahorias")
    plt.xlabel("Peso en gramos")
    plt.ylabel("Número de zanahorias")
    plt.grid(axis="y", linestyle="--", alpha=0.5)
    plt.show()

else:
    # Mensaje si no se introdujo ningún peso
    print("No has registrado ninguna zanahoria.")
```
