```python
# ==========================================
# 💻 MENÚ DE LENGUAJES DE PROGRAMACIÓN
# ==========================================

# Lista de lenguajes de programación
lenguajes = ["Python", "JavaScript", "Java", "C++", "PHP"]


# ------------------------------------------
# FUNCIÓN 1: Mostrar lenguajes
# ------------------------------------------
def mostrar_lenguajes():

    print("\n💻 LENGUAJES DISPONIBLES:")

    # Recorremos la lista
    for lenguaje in lenguajes:
        print("-", lenguaje)


# ------------------------------------------
# FUNCIÓN 2: Buscar un lenguaje
# ------------------------------------------
def buscar_lenguaje():

    # Pedimos un lenguaje al usuario
    nombre = input("\n¿Qué lenguaje quieres buscar? ")

    # Comprobamos si existe
    if nombre in lenguajes:
        print("✅ El lenguaje está en la lista.")
    else:
        print("❌ El lenguaje no está en la lista.")


# ------------------------------------------
# FUNCIÓN 3: Saber para qué sirve
# ------------------------------------------
def utilidad_lenguaje():

    nombre = input("\n¿Qué lenguaje quieres consultar? ")

    # Condiciones para cada lenguaje
    if nombre == "Python":
        print("🐍 Python se utiliza en IA, análisis de datos, automatización y desarrollo web.")

    elif nombre == "JavaScript":
        print("🌐 JavaScript se utiliza principalmente para crear páginas web interactivas.")

    elif nombre == "Java":
        print("☕ Java se utiliza en aplicaciones empresariales, Android y sistemas grandes.")

    elif nombre == "C++":
        print("⚙️ C++ se utiliza en videojuegos, sistemas y aplicaciones de alto rendimiento.")

    elif nombre == "PHP":
        print("🐘 PHP se utiliza principalmente para desarrollo web en el servidor.")

    else:
        print("❌ No tenemos información sobre ese lenguaje.")


# ------------------------------------------
# FUNCIÓN 4: Añadir un lenguaje
# ------------------------------------------
def añadir_lenguaje():

    nuevo = input("\nEscribe el nombre del nuevo lenguaje: ")

    # Comprobamos que no esté repetido
    if nuevo in lenguajes:
        print("⚠️ Ese lenguaje ya existe.")

    else:
        # Añadimos el lenguaje a la lista
        lenguajes.append(nuevo)

        print("✅ Lenguaje añadido correctamente.")


# ==========================================
# MENÚ PRINCIPAL
# ==========================================

# while True mantiene el menú funcionando
while True:




    print("💻 MENÚ DE PROGRAMACIÓN")
    print("==============================")
    print("1. Ver lenguajes")
    print("2. Buscar un lenguaje")
    print("3. Ver para qué sirve")
    print("4. Añadir un lenguaje")
    print("5. Salir")
    print("==============================")

    # Pedimos una opción
    opcion = input("Elige una opción: ")

    # Comprobamos la opción elegida
    if opcion == "1":
        mostrar_lenguajes()

    elif opcion == "2":
        buscar_lenguaje()

    elif opcion == "3":
        utilidad_lenguaje()

    elif opcion == "4":
        añadir_lenguaje()

    elif opcion == "5":
        print("\n👋 ¡Hasta luego!")
        break

    else:
        print("\n❌ Opción incorrecta. Elige entre 1 y 5.")
```
