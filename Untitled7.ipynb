Sistema Inteligente de Recomendación Musical


Crear un pequeño sistema en Python que pregunte al usuario cómo se siente y, según su estado de ánimo, le recomiende canciones de forma aleatoria.
Este ejercicio introduce listas, diccionarios, funciones y la librería random.

```python
# ==========================================
# 🎵 PYTHON DJ - RECOMENDADOR MUSICAL
# ==========================================

# Importamos la librería random de Python.
# Nos permitirá elegir canciones de forma aleatoria.
import random


# Creamos una función llamada "mostrar_playlist".
# La función recibirá el estado de ánimo elegido por el usuario.
def mostrar_playlist(estado):

    # Creamos un diccionario llamado "playlists".
    # Un diccionario permite guardar información usando claves y valores.
    playlists = {

        # La clave "1" representa el estado de ánimo feliz.
        "1": {

            # Guardamos el nombre del estado de ánimo.
            "estado": "😊 Feliz",

            # Guardamos un mensaje que mostraremos al usuario.
            "mensaje": "¡Perfecto! Vamos a mantener esa energía.",

            # Creamos una lista con canciones para este estado.
            "canciones": [
                "Don't Start Now - Dua Lipa",
                "Happy - Pharrell Williams",
                "Uptown Funk - Mark Ronson ft. Bruno Mars",
                "On Top of the World - Imagine Dragons",
                "Good as Hell - Lizzo"
            ]
        },

        # La clave "2" representa el estado triste.
        "2": {

            # Nombre del estado.
            "estado": "😢 Triste",

            # Mensaje que verá el usuario.
            "mensaje": "A veces una buena canción ayuda a cambiar el ánimo.",

            # Lista de canciones para este estado.
            "canciones": [
                "Someone Like You - Adele",
                "Lovely - Billie Eilish",
                "Fix You - Coldplay",
                "The Night We Met - Lord Huron",
                "When I Was Your Man - Bruno Mars"
            ]
        },

        # La clave "3" representa el estado relajado.
        "3": {

            # Nombre del estado.
            "estado": "😴 Relajado",

            # Mensaje para el usuario.
            "mensaje": "Perfecto para desconectar un poco.",

            # Canciones recomendadas.
            "canciones": [
                "Ocean Eyes - Billie Eilish",
                "Yellow - Coldplay",
                "Until I Found You - Stephen Sanchez",
                "Lovely - Billie Eilish",
                "Perfect - Ed Sheeran"
            ]
        },

        # La clave "4" representa el estado motivado.
        "4": {

            # Nombre del estado.
            "estado": "🔥 Motivado",

            # Mensaje para el usuario.
            "mensaje": "¡Vamos! Es hora de darlo todo.",

            # Canciones para aumentar la motivación.
            "canciones": [
                "Believer - Imagine Dragons",
                "Eye of the Tiger - Survivor",
                "Titanium - David Guetta",
                "Stronger - Kanye West",
                "Can't Hold Us - Macklemore"
            ]
        },

        # La clave "5" representa el estado enamorado.
        "5": {

            # Nombre del estado.
            "estado": "❤️ Enamorado",

            # Mensaje para el usuario.
            "mensaje": "Activando el modo romántico...",

            # Canciones románticas.
            "canciones": [
                "Perfect - Ed Sheeran",
                "All of Me - John Legend",
                "A Thousand Years - Christina Perri",
                "Thinking Out Loud - Ed Sheeran",
                "Just the Way You Are - Bruno Mars"
            ]
        }
    }


    # Comprobamos si la opción introducida existe en nuestro diccionario.
    if estado in playlists:

        # Guardamos toda la información del estado elegido
        # dentro de una variable llamada "playlist".
        playlist = playlists[estado]


        # Mostramos un mensaje indicando que estamos procesando la información.
        print("\n🎧 ANALIZANDO TU ESTADO DE ÁNIMO...")

        # Simulamos una barra de progreso.
        print("████████████████████ 100%")


        # Imprimimos una línea para separar visualmente la información.
        print("\n" + "=" * 45)

        # Mostramos el nombre del programa.
        print("       🎵 PYTHON DJ 🎵")

        # Imprimimos otra línea de separación.
        print("=" * 45)


        # Mostramos el estado de ánimo que ha seleccionado el usuario.
        # playlist["estado"] obtiene el valor guardado en la clave "estado".
        print(f"\nEstado detectado: {playlist['estado']}")


        # Mostramos el mensaje relacionado con el estado de ánimo.
        print(f"💬 {playlist['mensaje']}")


        # Seleccionamos 3 canciones aleatorias de la lista.
        #
        # random.sample() permite seleccionar varios elementos
        # sin repetirlos.
        recomendaciones = random.sample(
            playlist["canciones"],
            3
        )


        # Mostramos un título antes de las canciones.
        print("\n🎶 TU PLAYLIST PERSONALIZADA:\n")


        # enumerate() nos permite obtener:
        # 1. El número de la canción.
        # 2. El nombre de la canción.
        #
        # start=1 hace que la numeración empiece en 1.
        for numero, cancion in enumerate(recomendaciones, 1):

            # Mostramos el número y el nombre de cada canción.
            print(f"🎵 {numero}. {cancion}")


        # Mensaje final.
        print("\n🤖 ¡Playlist creada por Python DJ!")


    # Si el usuario introduce algo que no existe...
    else:

        # Mostramos un mensaje de error.
        print("\n❌ Opción no válida.")


# ==========================================
# PROGRAMA PRINCIPAL
# ==========================================


# Mostramos una línea decorativa.
print("=" * 45)

# Mostramos el título del programa.
print("        🎵 PYTHON DJ 🎵")

# Mostramos otra línea decorativa.
print("=" * 45)


# Preguntamos al usuario cómo se siente.
print("\n¿Cómo te sientes hoy?\n")


# Mostramos las diferentes opciones.
print("1️⃣ 😊 Feliz")
print("2️⃣ 😢 Triste")
print("3️⃣ 😴 Relajado")
print("4️⃣ 🔥 Motivado")
print("5️⃣ ❤️ Enamorado")


# input() permite que el usuario escriba una opción.
# Lo que escriba se guarda dentro de la variable "estado".
estado = input("\nSelecciona una opción: ")


# Llamamos a nuestra función "mostrar_playlist".
# Le pasamos la opción que introdujo el usuario.
mostrar_playlist(estado)

```
